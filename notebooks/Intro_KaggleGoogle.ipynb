## Ejemplo de conexión a un Dataset de Kaggle

Normalmente en todos los dataset de Kaggle en la parte superior tienen un botón que dice `Code` ahí nos muestra como debemos conectarnos al dataset desde un programa en Python.

### Método 1

```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
import kagglehub
from kagglehub import KaggleDatasetAdapter
```

```python
#df = pd.read_csv('/kaggle/input/datasets/abdulmalik1518/cars-datasets-2025/Cars Datasets 2025.csv',encoding='latin-1')
```

```python
# Set the path to the file you'd like to load
file_path = "Cars Datasets 2025.csv"  # <-- Este nombre se puede visualizar en el lado derecho del sitio en Kaggle, se debe indicar el que se desea.

# Load the latest version
df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "abdulmalik1518/cars-datasets-2025",
  file_path,
  pandas_kwargs={"encoding": "latin-1"},   # <-- Si el dataset está en una codificación diferente a UTF-8 debemos indicarlo
  # Provide any additional arguments like
  # sql_query or pandas_kwargs. See the
  # documenation for more information:
  # https://github.com/Kaggle/kagglehub/blob/main/README.md#kaggledatasetadapterpandas
)

print("First 5 records:", df.head())
```

### Método 2


Para este método se debe crear un API_TOKEN en la configuración de la cuenta de Kaggle. Posteriormente agregarlo en Colab usando el botón de `Secretos` de la barra de la izquierda (es una llave).

```python
import os
from google.colab import userdata
import kagglehub

# 1. Recuperamos el token guardado de forma segura en Colab
os.environ["KAGGLE_API_TOKEN"] = userdata.get('KAGGLE_API_TOKEN')

# 2. Descargamos el dataset usando la nueva herramienta kagglehub
# Reemplaza con el usuario/nombre-del-dataset que busques
path = kagglehub.dataset_download("abdulmalik1518/cars-datasets-2025")

print("El dataset se descargó correctamente en la ruta:", path)
```

## Ejemplo de descarga de archivo desde Google Drive

```python
# Instalamos la biblioteca gdown

!pip install gdown
```

Para descargar el archivo desde Google Drive, se debe haber compartido con derechos de [Lector] y vanos a utilizar el código del archivo el cual podemso extraer del vínculo de compartición.

Ejemplo:

https://drive.google.com/file/d/1YrTSVK2NyYAVGn1rOLz8HQlsvFEY8oqD/view?usp=sharing

En el vínculo anterior el código está entre las secciones `https://drive.google.com/file/d/` y `/view?usp=sharing`

```python
import gdown
#En el parámetro output le indicamos el nombre del archivo con el que se guardará.
gdown.download(id='1YrTSVK2NyYAVGn1rOLz8HQlsvFEY8oqD', output='nombre_archivo.csv', quiet=False)
```
